```python
#  Array Creation Functions
# 1]  BASIC :
```

```python
#  list into ndarray
#  1)
import numpy as np

a = np.array([1,2,0],dtype=float)


print(a)
```

```python
# 2) zeros and ones 
c=np.zeros((3,3))
d=np.ones((4,2))
print("zeros",c)
print("ones",d)

```

```python
# 3) np.full()
n=3
m=2
d=np.full((n,m), 47)  # fills n x m array with 7

print(d)
```

```python
# 4) np.eye()
e=np.eye(3)  # 3x3 identity matrix
print(e)
```

```python
# 5) np.empty()
f=np.empty((2,3))
print(f)
```

```python
# 2] Range 

# 1) np.arange()  act like range() function 
x=np.arange(0, 10, 2)
print(x)
y=np.arange(5)     
print(y)
```

```python
x=np.linspace(500,1000,5)     # (b-a)/n-1
print(x)
```

```python
t=np.logspace(1, 3, 4)  
print(t)
```

```python
# 3] Random 
# 1) np.random.rand()   
arr=np.random.rand(3,3)    # generate random num between 0-1 float 
print(arr)
```

```python
# 2) 
ab=np.random.randn(2,2)
print(ab)
```

```python
# 3)
p=np.random.randint(0,10,(3,3))
print(p)
q=np.random.randint(99)   # one num as parameter means [0,n) and one element 
print(q)
```

```python
# 4)
np.random.seed(42)
#  random ness become constant , and argument is any nums u put 
a = np.random.rand(4)      
b = np.random.rand(2, 4)   

print(a)  
print(b)  
```

```python
arr = np.array([10, 20, 30, 40, 50])
result = np.random.choice(arr, 3)
print(result)     
```

```python
# 5) np.random.shuffle() --> operate on mutabilities 
v=[9,23,44,65,87,92]
print(v)
np.random.shuffle(v)
print(v)
```

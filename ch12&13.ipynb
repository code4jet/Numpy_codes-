```python
# Copying & Memory
import numpy as np
# arr.view()
# What it does: Creates new array object but shares same memory
# Example:
arr = np.array([1, 2, 3, 4])
v = arr.view()

v[0] = 99
print(arr)  # [99,  2,  3,  4]  ← original changed!
```

```python
# arr.copy()
# What it does: Creates new array with new memory
# Example:
arr = np.array([1, 2, 3, 4])
c = arr.copy()

c[0] = 99
print(arr)  # [1, 2, 3, 4]  ← original safe!
print(c)    # [99, 2, 3, 4]
```

```python
# arr.base ( bool output  )
# What it does: Tells you who owns the memory
arr = np.array([1, 2, 3, 4])
v = arr.view()
c = arr.copy()

print(v.base is arr)  # True  ← view depends on arr
print(c.base is arr)  # False ← copy is independent
```

```python
# MEMORY Functions   
# arr.flags
arr = np.array([1, 2, 3, 4])
print(arr.flags)
# OWNDATA = True   ← owns its memory
v = arr.view()
print(v.flags)
# OWNDATA = False  ← borrowed memory
```

```python
# np.shares_memory(a, b)
arr = np.array([1, 2, 3])
v = arr.view()
c = arr.copy()

print(np.shares_memory(arr, v) ) # True
print(np.shares_memory(arr, c) ) # False

# np.may_share_memory(a, b)
print(np.may_share_memory(arr, v) ) # True  (quick check)
# Difference:
# shares_memory()     → exact check   → slow
# may_share_memory()  → quick check   → fast (used in large arrays)
```

```python
#  Vectorization & Performance Important
# VECTORIZE
# np.vectorize() np.frompyfunc() np.apply_along_axis()
# BEST PRACTICE
# avoid Python loops use broadcasting
# instead
# use functions

def sq(x):
    return x*x
x=np.array([4,5,6,6,7,8])
z=np.vectorize(sq)
print(z(x))

# np.vectorize() is still a loop internally
# cleaner code but NOT faster than manual loop
# use it for convenience, not speed
```

```python
# np.frompyfunc(func, nin, nout)
# What it does: Like vectorize but returns object array, lower level
# Example:
def add(x, y):
    return x + y

ufunc_add = np.frompyfunc(add, 2, 1)
#                              ↑  ↑
#                         inputs outputs

arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

print(ufunc_add(arr1, arr2))  # [5, 7, 9]

# nin  = 2 → function takes 2 inputs
# nout = 1 → function gives 1 output

# ufunc_add([1,2,3], [4,5,6])
# → [add(1,4), add(2,5), add(3,6)]
# → [5, 7, 9]
```

```python
# np.apply_along_axis(func, axis, arr)
# What it does: Apply function along row or column
# Example:
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

result = np.apply_along_axis(np.sum, axis=1, arr=arr)
print(result)  # [6, 15]
```

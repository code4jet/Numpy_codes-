float32 vs float64 matters a lot in deep learning memory usage.


```python
x=np.array([5,9,45])
y=np.array([5.6,7.983,1,243])
z=np.array([True,False])
print(x.dtype)
print(y.dtype)
print(z.dtype)
a = np.array([1,2,3], dtype=np.int8)
print(a.size)

print(a.dtype)
```

```python
n=np.array([3,4])
m=np.array([
    [4,5,7,8,9],
    [8,2,3,0,8]
])
print("n: ",n.itemsize, n.nbytes  )       # itemsize : give memory is used by one element 
print("m: ",m.itemsize, m.nbytes  )       # nbytes   : give memory is used by one element 
```
